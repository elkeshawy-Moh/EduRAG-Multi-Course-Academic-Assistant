# STEP 4: MULTI-FORMAT DOCUMENT LOADING, CLEANING & METADATA ENRICHMENT

````python
# ==============================================================================
# STEP 4: MULTI-FORMAT DOCUMENT LOADING, CLEANING & METADATA ENRICHMENT
# ==============================================================================
import os
import re
from langchain_community.document_loaders import (
    PyMuPDFLoader,
    Docx2txtLoader,
    CSVLoader,
    TextLoader
)

def clean_text(text: str) -> str:
    """
    Cleans extracted text by normalizing whitespace and removing redundant newlines
    while preserving paragraph structure.
    """
    # Replace multiple spaces/tabs with a single space
    text = re.sub(r'[ \t]+', ' ', text)
    # Replace 3 or more consecutive newlines with 2 newlines
    text = re.sub(r'\n{3,}', '\n\n', text)
    return text.strip()

def load_all_courses(data_directory: str):
    """
    Iterates through all course folders, selects the appropriate loader based on
    file extension (PDF, DOCX, CSV, TXT), cleans the text, and enriches metadata.
    """
    all_documents = []

    # Iterate over each course directory
    for course_name in sorted(os.listdir(data_directory)):
        course_path = os.path.join(data_directory, course_name)
        if not os.path.isdir(course_path):
            continue

        print(f"\n📂 Processing Course: {course_name}")

        for file_name in sorted(os.listdir(course_path)):
            file_path = os.path.join(course_path, file_name)
            file_ext = os.path.splitext(file_name)[1].lower()

            # Select specialized loader for each file format (not just PyPDF)
            try:
                if file_ext == ".pdf":
                    loader = PyMuPDFLoader(file_path)
                elif file_ext == ".docx":
                    loader = Docx2txtLoader(file_path)
                elif file_ext == ".csv":
                    loader = CSVLoader(file_path, encoding="utf-8")
                elif file_ext == ".txt":
                    loader = TextLoader(file_path, encoding="utf-8")
                else:
                    print(f"  ⚠️ Skipping unsupported file format: {file_name}")
                    continue

                # Load documents from the file
                docs = loader.load()

                # Clean text and attach rich metadata to every document
                for doc in docs:
                    doc.page_content = clean_text(doc.page_content)

                    # Preserve original page number if available (1-indexed for readability)
                    raw_page = doc.metadata.get("page", None)
                    page_num = (raw_page + 1) if isinstance(raw_page, int) else "N/A"

                    # Update metadata dictionary for accurate source attribution later
                    doc.metadata.update({
                        "course": course_name,
                        "file_name": file_name,
                        "file_type": file_ext.replace(".", "").upper(),
                        "page": page_num
                    })

                all_documents.extend(docs)
                print(f"  ✅ Loaded '{file_name}' ({file_ext.upper()}) -> {len(docs)} document unit(s)")

            except Exception as e:
                print(f"  ❌ Error loading {file_name}: {str(e)}")

    return all_documents

# Execute the multi-format loader
raw_documents = load_all_courses(DATA_DIR)
print(f"\n📊 Total raw document units loaded across all courses: {len(raw_documents)}")
````

# STEP 5: TWO-STAGE CHUNKING (RECURSIVE CHARACTER + TOKEN SPLITTER WITH OVERLAP)

````python
# ==============================================================================
# STEP 5: TWO-STAGE CHUNKING (RECURSIVE CHARACTER + TOKEN SPLITTER WITH OVERLAP)
# ==============================================================================
from langchain_text_splitters import RecursiveCharacterTextSplitter, TokenTextSplitter

def split_documents_hybrid(documents):
    """
    Applies a two-stage splitting strategy:
    1. RecursiveCharacterTextSplitter: Preserves semantic boundaries (paragraphs, sentences).
    2. TokenTextSplitter: Enforces strict token limits compatible with embedding models.
    """
    # Stage 1: Semantic splitting using RecursiveCharacterTextSplitter
    # Separators are ordered from largest structural unit to smallest
    recursive_splitter = RecursiveCharacterTextSplitter(
        chunk_size=400,
        chunk_overlap=80,  # 20% overlap to prevent losing context at boundaries
        separators=["\n\n", "\n", ". ", " ", ""],
        length_function=len
    )
    recursive_chunks = recursive_splitter.split_documents(documents)

    # Stage 2: Token-level enforcement using TokenTextSplitter (via tiktoken)
    # Ensures no chunk exceeds the token window of the embedding model
    token_splitter = TokenTextSplitter(
        encoding_name="cl100k_base",
        chunk_size=120,
        chunk_overlap=20
    )
    final_chunks = token_splitter.split_documents(recursive_chunks)

    # Assign a unique chunk index and enrich chunk text with source header
    for idx, chunk in enumerate(final_chunks):
        chunk.metadata["chunk_id"] = f"chunk_{idx:03d}"

    return recursive_chunks, final_chunks

# Run the hybrid splitting pipeline
stage1_chunks, final_chunks = split_documents_hybrid(raw_documents)

print(f"🔹 After Stage 1 (Recursive Splitter): {len(stage1_chunks)} chunks")
print(f"🔹 After Stage 2 (Token Splitter):     {len(final_chunks)} final chunks\n")

# Inspect a sample chunk to verify text quality and preserved metadata
print("🔍 Sample Chunk Inspection:")
print("-" * 60)
print(f"Metadata: {final_chunks[0].metadata}")
print(f"Content:\n{final_chunks[0].page_content}")
print("-" * 60)
````

# STEP 6: INITIALIZE EMBEDDING MODEL & BUILD PERSISTENT CHROMA VECTOR DATABASE

````python
# ==============================================================================
# STEP 6: INITIALIZE EMBEDDING MODEL & BUILD PERSISTENT CHROMA VECTOR DATABASE
# ==============================================================================
# Install langchain-chroma first so it is available for import
!pip install -q langchain-chroma

import shutil
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
import os

# 1. Initialize the Embedding Model
# CRITICAL RULE: The exact same embedding model MUST be used for both
# Indexing (storing chunks) and Querying (searching user questions).
EMBEDDING_MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"

print(f"⏳ Loading Embedding Model: {EMBEDDING_MODEL_NAME}...")
embedding_model = HuggingFaceEmbeddings(
    model_name=EMBEDDING_MODEL_NAME,
    model_kwargs={"device": "cpu"},  # Use "cuda" if GPU is enabled in Colab
    encode_kwargs={"normalize_embeddings": True}  # Normalize vectors for accurate cosine similarity
)
print("✅ Embedding Model loaded successfully!")

# 2. Clean previous database directory if re-running to avoid duplicate entries
if os.path.exists(DB_DIR):
    shutil.rmtree(DB_DIR)
os.makedirs(DB_DIR, exist_ok=True)

# 3. Create and persist the Chroma Vector Database from our final chunks
print("⏳ Indexing chunks into ChromaDB...")
vector_db = Chroma.from_documents(
    documents=final_chunks,
    embedding=embedding_model,
    persist_directory=DB_DIR,
    collection_name="edurag_courses_collection"
)

print(f"✅ ChromaDB built and persisted at '{DB_DIR}'!")
print(f"📦 Total vectors stored in database: {vector_db._collection.count()}")
````



# STEP 9: CONTEXT ENRICHMENT, STRICT PROMPTING & END-TO-END RAG PIPELINE


````python
# ==============================================================================
# STEP 9 (CORRECTED): CONTEXT ENRICHMENT, STRICT PROMPTING & END-TO-END RAG
# ==============================================================================
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Verify that 'llm' from Step 8 is initialized before running
if "llm" not in globals() or llm is None:
    raise RuntimeError("❌ 'llm' is not initialized. Please run Step 8 first!")

# 1. Define the strict System & User Prompt Template
RAG_PROMPT_TEMPLATE = """
You are 'EduRAG', an authoritative and helpful Multi-Course Academic Assistant.
Your task is to answer student questions strictly and exclusively based on the provided Course Context below.

STRICT RULES:
1. Ground your answer ONLY in the facts explicitly stated in the provided context. Do NOT use outside knowledge.
2. NEVER say "I don't know". If the provided context does not contain sufficient evidence to answer the question, respond with the exact phrase:
   "I don't have enough proof in the provided course materials to answer this question."
3. Maintain a confident, clear, and academic tone.
4. Cite the exact source file name(s) and page number(s) inline or at the end of your explanation.

--- COURSE CONTEXT ---
{enriched_context}
----------------------

Student Question: {question}

Verified Academic Answer:
"""

prompt_template = ChatPromptTemplate.from_template(RAG_PROMPT_TEMPLATE)

def format_and_enrich_context(retrieved_docs) -> tuple[str, list[str]]:
    """
    Enriches retrieved chunks by prepending structured metadata headers
    so the LLM knows the exact course, file, and page for every snippet.
    Also extracts a clean list of unique source citations.
    """
    enriched_blocks = []
    unique_sources = []

    for idx, doc in enumerate(retrieved_docs, start=1):
        meta = doc.metadata
        course = meta.get("course", "Unknown Course")
        file_name = meta.get("file_name", "Unknown File")
        file_type = meta.get("file_type", "DOC")
        page = meta.get("page", "N/A")
        chunk_id = meta.get("chunk_id", "N/A")

        # Build enriched header for the LLM context
        header = f"[Snippet {idx} | Course: {course} | File: {file_name} ({file_type}) | Page: {page} | ID: {chunk_id}]"
        enriched_blocks.append(f"{header}\n{doc.page_content}")

        # Format user-facing citation string
        citation = f"📘 **{course}** — `{file_name}` (Page: {page}, {chunk_id})"
        if citation not in unique_sources:
            unique_sources.append(citation)

    return "\n\n".join(enriched_blocks), unique_sources

def ask_edurag(question: str, selected_course: str = "ALL", top_k: int = 3) -> dict:
    """
    Executes the full Standard RAG pipeline:
    1. Retrieves relevant chunks (with optional course filter).
    2. Enriches context with metadata.
    3. Generates a grounded answer with source attribution.
    """
    if not question.strip():
        return {
            "answer": "Please enter a valid academic question.",
            "sources": []
        }

    # Configure retriever with optional metadata filtering
    search_kwargs = {"k": top_k}
    if selected_course != "ALL":
        search_kwargs["filter"] = {"course": selected_course}

    retriever = vector_db.as_retriever(
        search_type="similarity",
        search_kwargs=search_kwargs
    )

    # Step A: Retrieve matching chunks
    retrieved_docs = retriever.invoke(question)

    # Fallback if database returns nothing
    if not retrieved_docs:
        return {
            "answer": "I don't have enough proof in the provided course materials to answer this question.",
            "sources": []
        }

    # Step B: Enrich context with metadata
    enriched_context, sources_list = format_and_enrich_context(retrieved_docs)

    # Step C: Build and invoke the LangChain LCEL pipeline
    rag_chain = prompt_template | llm | StrOutputParser()
    answer = rag_chain.invoke({
        "enriched_context": enriched_context,
        "question": question
    })

    # If the model states there is not enough proof, clear irrelevant sources
    if "don't have enough proof" in answer.lower():
        sources_list = []

    return {
        "answer": answer.strip(),
        "sources": sources_list
    }

# ==============================================================================
# TEST THE END-TO-END RAG PIPELINE (IN-DOMAIN VS. OUT-OF-DOMAIN QUESTIONS)
# ==============================================================================
print("🧪 TEST 1: Question inside course materials (Software Engineering)")
res1 = ask_edurag(
    "What are the three key roles in Scrum and what does the Scrum Master do?",
    selected_course="SE303_Software_Engineering"
)
print("Answer:\n", res1["answer"])
print("\nSources:\n", "\n".join(res1["sources"]))
print("=" * 75)

print("\n🧪 TEST 2: Question completely outside course materials (Hallucination Check)")
res2 = ask_edurag("How do I bake a chocolate cake?", selected_course="ALL")
print("Answer:\n", res2["answer"])
print("\nSources:", res2["sources"])
print("=" * 75)
````

# STEP 10: BUILD & LAUNCH PROFESSIONAL GRADIO INTERFACE FOR EDURAG

````python
# ==============================================================================
# STEP 10 (GRADIO 6.0 COMPATIBLE): BUILD & LAUNCH PROFESSIONAL GRADIO UI
# ==============================================================================
import gradio as gr
import shutil
import os

def gradio_chat_handler(message, history, selected_course, top_k):
    """
    Handles user queries from the Gradio Chatbot UI, invokes the RAG pipeline,
    and returns updated chat history alongside formatted source citations.
    """
    if not message.strip():
        return history, "⚠️ Please enter a question first."

    # Query the Standard RAG pipeline
    result = ask_edurag(
        question=message,
        selected_course=selected_course,
        top_k=int(top_k)
    )

    answer = result["answer"]
    sources = result["sources"]

    # Format sources panel output
    if sources:
        sources_markdown = "### 📚 Verified Source Attribution\n" + "\n".join([f"- {src}" for src in sources])
    else:
        sources_markdown = "### 🛡️ Grounding Status\nNo supporting proof found in the selected course materials."

    # Append interaction to chat history (Gradio 6.0 native messages format)
    history = history or []
    history.append({"role": "user", "content": message})
    history.append({"role": "assistant", "content": answer})

    return history, sources_markdown

def handle_file_upload(uploaded_file, target_course):
    """
    Allows dynamic uploading of new course files (PDF, DOCX, CSV, TXT) via UI,
    processes them with the hybrid splitter, and adds them to ChromaDB.
    """
    if uploaded_file is None:
        return "⚠️ Please select a file to upload."

    file_name = os.path.basename(uploaded_file.name)
    file_ext = os.path.splitext(file_name)[1].lower()

    if file_ext not in [".pdf", ".docx", ".csv", ".txt"]:
        return f"❌ Unsupported file type `{file_ext}`. Allowed: PDF, DOCX, CSV, TXT."

    # Save uploaded file into the target course directory
    dest_path = os.path.join(DATA_DIR, target_course, file_name)
    shutil.copy(uploaded_file.name, dest_path)

    # Load, clean, split, and index the new file
    try:
        if file_ext == ".pdf":
            loader = PyMuPDFLoader(dest_path)
        elif file_ext == ".docx":
            loader = Docx2txtLoader(dest_path)
        elif file_ext == ".csv":
            loader = CSVLoader(dest_path, encoding="utf-8")
        elif file_ext == ".txt":
            loader = TextLoader(dest_path, encoding="utf-8")

        docs = loader.load()
        for doc in docs:
            doc.page_content = clean_text(doc.page_content)
            raw_page = doc.metadata.get("page", None)
            page_num = (raw_page + 1) if isinstance(raw_page, int) else "N/A"
            doc.metadata.update({
                "course": target_course,
                "file_name": file_name,
                "file_type": file_ext.replace(".", "").upper(),
                "page": page_num
            })

        _, new_chunks = split_documents_hybrid(docs)
        vector_db.add_documents(new_chunks)

        return (
            f"✅ Successfully indexed **`{file_name}`** into **{target_course}**!\n"
            f"- Added **{len(new_chunks)}** new chunks to ChromaDB.\n"
            f"- Total vectors in knowledge base: **{vector_db._collection.count()}**"
        )
    except Exception as e:
        return f"❌ Error indexing file: {str(e)}"

# Define available course filter options
COURSE_OPTIONS = ["ALL"] + COURSES

# Build the Gradio Blocks UI (theme moved to launch() for Gradio 6.0)
with gr.Blocks(title="EduRAG: Multi-Course Academic Assistant") as demo:

    # Project Header Banner
    gr.Markdown(
        """
        # 🎓 EduRAG: Multi-Course Academic Assistant
        **Standard Retrieval-Augmented Generation (RAG) System** | Grounded strictly in verified course materials (`PDF`, `DOCX`, `CSV`, `TXT`).
        """
    )

    with gr.Tabs():
        # ======================================================================
        # TAB 1: COURSE Q&A ASSISTANT
        # ======================================================================
        with gr.TabItem("💬 Course Q&A Assistant"):
            with gr.Row():
                # Left Column: Controls & Filters
                with gr.Column(scale=1):
                    gr.Markdown("### ⚙️ Retrieval Settings")
                    course_dropdown = gr.Dropdown(
                        choices=COURSE_OPTIONS,
                        value="ALL",
                        label="Filter by Course",
                        info="Select 'ALL' to search across all 3 courses, or isolate a single course."
                    )
                    topk_slider = gr.Slider(
                        minimum=1,
                        maximum=6,
                        value=3,
                        step=1,
                        label="Top-K Retrieved Chunks",
                        info="Number of context snippets passed to the LLM."
                    )
                    clear_btn = gr.Button("🗑️ Clear Conversation", variant="secondary")

                    gr.Markdown("### 💡 Quick Test Prompts")
                    ex1_btn = gr.Button("📌 Scrum Roles (SE303)", size="sm")
                    ex2_btn = gr.Button("📌 Third Normal Form 3NF (DB202)", size="sm")
                    ex3_btn = gr.Button("📌 AI Midterm Policy (CS101)", size="sm")
                    ex4_btn = gr.Button("🚫 Out-of-Scope Test (No Proof)", size="sm")

                # Middle Column: Main Chat Interface
                with gr.Column(scale=2):
                    chatbot = gr.Chatbot(
                        label="Academic Assistant Chat",
                        height=440
                    )
                    with gr.Row():
                        msg_input = gr.Textbox(
                            placeholder="Ask a question from your course lectures, syllabi, or notes...",
                            label="Your Question",
                            scale=4,
                            show_label=False
                        )
                        send_btn = gr.Button("Ask 🚀", variant="primary", scale=1)

                # Right Column: Live Source Citations Panel
                with gr.Column(scale=1):
                    sources_panel = gr.Markdown(
                        value="### 📚 Verified Source Attribution\nAsk a question to view exact file names, page numbers, and chunk IDs."
                    )

            # Event Bindings for Chat
            send_btn.click(
                fn=gradio_chat_handler,
                inputs=[msg_input, chatbot, course_dropdown, topk_slider],
                outputs=[chatbot, sources_panel]
            ).then(lambda: "", None, msg_input)

            msg_input.submit(
                fn=gradio_chat_handler,
                inputs=[msg_input, chatbot, course_dropdown, topk_slider],
                outputs=[chatbot, sources_panel]
            ).then(lambda: "", None, msg_input)

            clear_btn.click(
                fn=lambda: ([], "### 📚 Verified Source Attribution\nConversation cleared."),
                inputs=None,
                outputs=[chatbot, sources_panel]
            )

            # Quick Example Buttons Bindings
            ex1_btn.click(lambda: ("What are the three key roles in Scrum?", "SE303_Software_Engineering"), None, [msg_input, course_dropdown])
            ex2_btn.click(lambda: ("Explain Third Normal Form (3NF) in databases.", "DB202_Database_Systems"), None, [msg_input, course_dropdown])
            ex3_btn.click(lambda: ("What is the attendance and midterm exam policy in CS101?", "CS101_Artificial_Intelligence"), None, [msg_input, course_dropdown])
            ex4_btn.click(lambda: ("What is the recipe for pasta carbonara?", "ALL"), None, [msg_input, course_dropdown])

        # ======================================================================
        # TAB 2: DYNAMIC KNOWLEDGE BASE UPLOADER
        # ======================================================================
        with gr.TabItem("📤 Upload Course Material"):
            gr.Markdown(
                """
                ### Add New Documents to ChromaDB
                Upload additional course materials (**PDF, DOCX, CSV, TXT**). They will be parsed, cleaned, split using our Hybrid Splitter, and indexed immediately.
                """
            )
            with gr.Row():
                file_input = gr.File(label="Select Course File (PDF, DOCX, CSV, TXT)")
                upload_course_dropdown = gr.Dropdown(
                    choices=COURSES,
                    value=COURSES[0],
                    label="Assign to Course"
                )
            upload_btn = gr.Button("⚙️ Process & Index Document", variant="primary")
            upload_status = gr.Markdown("")

            upload_btn.click(
                fn=handle_file_upload,
                inputs=[file_input, upload_course_dropdown],
                outputs=upload_status
            )

# Launch Gradio 6.0 with theme passed directly inside launch()
demo.launch(
    share=True,
    debug=False,
    theme=gr.themes.Soft(primary_hue="blue", secondary_hue="slate")
)
````

````python
# ==============================================================================
# STEP 12: GENERATE STANDALONE 'app.py' FOR GITHUB & HUGGING FACE SPACES
# ==============================================================================

"""
EduRAG: Multi-Course Academic Assistant
Standard Retrieval-Augmented Generation (RAG) Pipeline with Gradio 6.0 UI.
Supports multi-format course materials (PDF, DOCX, CSV, TXT), two-stage hybrid
chunking, ChromaDB vector persistence, metadata filtering, and strict grounding.
"""

import os
import re
import shutil
import gradio as gr

from langchain_community.document_loaders import (
    PyMuPDFLoader,
    Docx2txtLoader,
    CSVLoader,
    TextLoader,
)
from langchain_text_splitters import RecursiveCharacterTextSplitter, TokenTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# ==============================================================================
# 1. CONFIGURATION & DIRECTORY PATHS
# ==============================================================================
try:
    BASE_DIR = os.path.dirname(os.path.abspath(__file__))
except NameError:
    # Fallback when executed inside a Jupyter / Google Colab notebook cell
    BASE_DIR = "/content/EduRAG_Project" if os.path.exists("/content/EduRAG_Project") else os.getcwd()

DATA_DIR = os.path.join(BASE_DIR, "course_materials")
DB_DIR = os.path.join(BASE_DIR, "vector_db")


# ==============================================================================
# 2. TEXT CLEANING & MULTI-FORMAT DOCUMENT LOADING
# ==============================================================================
def clean_text(text: str) -> str:
    """Normalizes whitespace and redundant newlines while preserving paragraphs."""
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()


def load_single_file(file_path: str, course_name: str):
    """Selects a specialized loader per file extension and enriches metadata."""
    file_name = os.path.basename(file_path)
    file_ext = os.path.splitext(file_name)[1].lower()

    if file_ext == ".pdf":
        loader = PyMuPDFLoader(file_path)
    elif file_ext == ".docx":
        loader = Docx2txtLoader(file_path)
    elif file_ext == ".csv":
        loader = CSVLoader(file_path, encoding="utf-8")
    elif file_ext == ".txt":
        loader = TextLoader(file_path, encoding="utf-8")
    else:
        return []

    docs = loader.load()
    for doc in docs:
        doc.page_content = clean_text(doc.page_content)
        raw_page = doc.metadata.get("page", None)
        page_num = (raw_page + 1) if isinstance(raw_page, int) else "N/A"
        doc.metadata.update({
            "course": course_name,
            "file_name": file_name,
            "file_type": file_ext.replace(".", "").upper(),
            "page": page_num,
        })
    return docs


def load_all_courses(data_directory: str):
    """Loads all course materials across all course subdirectories."""
    all_documents = []
    for course_name in sorted(os.listdir(data_directory)):
        course_path = os.path.join(data_directory, course_name)
        if not os.path.isdir(course_path):
            continue
        for file_name in sorted(os.listdir(course_path)):
            file_path = os.path.join(course_path, file_name)
            try:
                all_documents.extend(load_single_file(file_path, course_name))
            except Exception as exc:
                print(f"[Warning] Failed to load {file_name}: {exc}")
    return all_documents


# ==============================================================================
# 3. TWO-STAGE HYBRID CHUNKING (RECURSIVE + TOKEN SPLITTERS WITH OVERLAP)
# ==============================================================================
def split_documents_hybrid(documents, start_index: int = 0):
    """
    Stage 1: RecursiveCharacterTextSplitter preserves semantic boundaries.
    Stage 2: TokenTextSplitter enforces strict token limits for embeddings.
    """
    recursive_splitter = RecursiveCharacterTextSplitter(
        chunk_size=400,
        chunk_overlap=80,
        separators=["\n\n", "\n", ". ", " ", ""],
        length_function=len,
    )
    recursive_chunks = recursive_splitter.split_documents(documents)

    token_splitter = TokenTextSplitter(
        encoding_name="cl100k_base",
        chunk_size=120,
        chunk_overlap=20,
    )
    final_chunks = token_splitter.split_documents(recursive_chunks)

    for idx, chunk in enumerate(final_chunks, start=start_index):
        chunk.metadata["chunk_id"] = f"chunk_{idx:03d}"

    return final_chunks


# ==============================================================================
# 4. INITIALIZE EMBEDDING MODEL & CHROMA VECTOR DATABASE
# ==============================================================================
print(f"Loading embedding model: {EMBEDDING_MODEL_NAME}...")
embedding_model = HuggingFaceEmbeddings(
    model_name=EMBEDDING_MODEL_NAME,
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True},
)

# Load existing persisted ChromaDB or build a new one from course_materials
vector_db = Chroma(
    persist_directory=DB_DIR,
    embedding_function=embedding_model,
    collection_name="edurag_courses_collection",
)

if vector_db._collection.count() == 0:
    print("Vector database is empty. Indexing course materials...")
    raw_docs = load_all_courses(DATA_DIR)
    if raw_docs:
        chunks = split_documents_hybrid(raw_docs)
        vector_db.add_documents(chunks)
        print(f"Indexed {len(chunks)} chunks into ChromaDB.")
else:
    print(f"Loaded existing ChromaDB with {vector_db._collection.count()} vectors.")


# ==============================================================================
# 5. LLM INITIALIZATION & STRICT GROUNDED RAG PIPELINE
# ==============================================================================
def get_active_llm(api_key: str):
    """Selects and verifies an active Groq LLM using the provided API key."""
    clean_key = (api_key or os.environ.get("GROQ_API_KEY", "")).strip()
    if not clean_key:
        raise ValueError("Missing GROQ_API_KEY. Please provide your Groq API Key.")

    os.environ["GROQ_API_KEY"] = clean_key
    for model_id in CANDIDATE_LLM_MODELS:
        try:
            candidate = ChatGroq(model=model_id, temperature=0.0, max_tokens=1024)
            candidate.invoke("ping")
            return candidate
        except Exception:
            continue
    raise RuntimeError("No Groq models are accessible with the provided API key.")


RAG_PROMPT_TEMPLATE = """
You are 'EduRAG', an authoritative and helpful Multi-Course Academic Assistant.
Your task is to answer student questions strictly and exclusively based on the provided Course Context below.

STRICT RULES:
1. Ground your answer ONLY in the facts explicitly stated in the provided context. Do NOT use outside knowledge.
2. NEVER say "I don't know". If the provided context does not contain sufficient evidence to answer the question, respond with the exact phrase:
   "I don't have enough proof in the provided course materials to answer this question."
3. Maintain a confident, clear, and academic tone.
4. Cite the exact source file name(s) and page number(s) inline or at the end of your explanation.

--- COURSE CONTEXT ---
{enriched_context}
----------------------

Student Question: {question}

Verified Academic Answer:
"""
prompt_template = ChatPromptTemplate.from_template(RAG_PROMPT_TEMPLATE)


def format_and_enrich_context(retrieved_docs):
    """Prepends structured metadata headers to retrieved snippets."""
    enriched_blocks = []
    unique_sources = []

    for idx, doc in enumerate(retrieved_docs, start=1):
        meta = doc.metadata
        course = meta.get("course", "Unknown Course")
        file_name = meta.get("file_name", "Unknown File")
        file_type = meta.get("file_type", "DOC")
        page = meta.get("page", "N/A")
        chunk_id = meta.get("chunk_id", "N/A")

        header = (
            f"[Snippet {idx} | Course: {course} | File: {file_name} ({file_type}) "
            f"| Page: {page} | ID: {chunk_id}]"
        )
        enriched_blocks.append(f"{header}\n{doc.page_content}")

        citation = f"📘 **{course}** — `{file_name}` (Page: {page}, {chunk_id})"
        if citation not in unique_sources:
            unique_sources.append(citation)

    return "\n\n".join(enriched_blocks), unique_sources


def ask_edurag(question: str, selected_course: str, top_k: int, api_key: str) -> dict:
    """Executes the end-to-end Standard RAG retrieval and grounded generation."""
    if not question.strip():
        return {"answer": "Please enter a valid academic question.", "sources": []}

    try:
        active_llm = get_active_llm(api_key)
    except Exception as exc:
        return {"answer": f"⚠️ Configuration Error: {exc}", "sources": []}

    search_kwargs = {"k": int(top_k)}
    if selected_course != "ALL":
        search_kwargs["filter"] = {"course": selected_course}

    retriever = vector_db.as_retriever(
        search_type="similarity",
        search_kwargs=search_kwargs,
    )
    retrieved_docs = retriever.invoke(question)

    if not retrieved_docs:
        return {
            "answer": "I don't have enough proof in the provided course materials to answer this question.",
            "sources": [],
        }

    enriched_context, sources_list = format_and_enrich_context(retrieved_docs)
    rag_chain = prompt_template | active_llm | StrOutputParser()
    answer = rag_chain.invoke({
        "enriched_context": enriched_context,
        "question": question,
    })

    if "don't have enough proof" in answer.lower():
        sources_list = []

    return {"answer": answer.strip(), "sources": sources_list}


# ==============================================================================
# 6. GRADIO 6.0 WEB APPLICATION INTERFACE
# ==============================================================================
def get_available_courses():
    courses = sorted([
        d for d in os.listdir(DATA_DIR)
        if os.path.isdir(os.path.join(DATA_DIR, d))
    ])
    return courses if courses else [
        "CS101_Artificial_Intelligence",
        "DB202_Database_Systems",
        "SE303_Software_Engineering",
    ]


def gradio_chat_handler(message, history, selected_course, top_k, api_key):
    if not message.strip():
        return history, "⚠️ Please enter a question first."

    result = ask_edurag(message, selected_course, top_k, api_key)
    answer = result["answer"]
    sources = result["sources"]

    if sources:
        sources_md = "### 📚 Verified Source Attribution\n" + "\n".join([f"- {s}" for s in sources])
    else:
        sources_md = "### 🛡️ Grounding Status\nNo supporting proof found in the selected course materials."

    history = history or []
    history.append({"role": "user", "content": message})
    history.append({"role": "assistant", "content": answer})
    return history, sources_md


def handle_file_upload(uploaded_file, target_course):
    if uploaded_file is None:
        return "⚠️ Please select a file to upload."

    file_name = os.path.basename(uploaded_file.name)
    file_ext = os.path.splitext(file_name)[1].lower()
    if file_ext not in [".pdf", ".docx", ".csv", ".txt"]:
        return f"❌ Unsupported file type `{file_ext}`. Allowed: PDF, DOCX, CSV, TXT."

    course_folder = os.path.join(DATA_DIR, target_course)
    os.makedirs(course_folder, exist_ok=True)
    dest_path = os.path.join(course_folder, file_name)
    shutil.copy(uploaded_file.name, dest_path)

    try:
        docs = load_single_file(dest_path, target_course)
        current_count = vector_db._collection.count()
        new_chunks = split_documents_hybrid(docs, start_index=current_count)
        vector_db.add_documents(new_chunks)
        return (
            f"✅ Successfully indexed **`{file_name}`** into **{target_course}**!\n"
            f"- Added **{len(new_chunks)}** new chunks to ChromaDB.\n"
            f"- Total vectors in knowledge base: **{vector_db._collection.count()}**"
        )
    except Exception as exc:
        return f"❌ Error indexing file: {exc}"


AVAILABLE_COURSES = get_available_courses()
COURSE_OPTIONS = ["ALL"] + AVAILABLE_COURSES

with gr.Blocks(title="EduRAG: Multi-Course Academic Assistant") as demo:
    gr.Markdown(
        """
        # 🎓 EduRAG: Multi-Course Academic Assistant
        **Standard Retrieval-Augmented Generation (RAG) System** | Grounded strictly in verified course materials (`PDF`, `DOCX`, `CSV`, `TXT`).
        """
    )

    with gr.Tabs():
        with gr.TabItem("💬 Course Q&A Assistant"):
            with gr.Row():
                with gr.Column(scale=1):
                    gr.Markdown("### ⚙️ Configuration & Filters")
                    api_key_box = gr.Textbox(
                        label="Groq API Key",
                        placeholder="gsk_... (or set GROQ_API_KEY env var)",
                        type="password",
                        value=os.environ.get("GROQ_API_KEY", ""),
                    )
                    course_dropdown = gr.Dropdown(
                        choices=COURSE_OPTIONS,
                        value="ALL",
                        label="Filter by Course",
                    )
                    topk_slider = gr.Slider(
                        minimum=1,
                        maximum=6,
                        value=3,
                        step=1,
                        label="Top-K Retrieved Chunks",
                    )
                    clear_btn = gr.Button("🗑️ Clear Conversation", variant="secondary")

                    gr.Markdown("### 💡 Quick Test Prompts")
                    ex1_btn = gr.Button("📌 Scrum Roles (SE303)", size="sm")
                    ex2_btn = gr.Button("📌 Third Normal Form 3NF (DB202)", size="sm")
                    ex3_btn = gr.Button("📌 AI Midterm Policy (CS101)", size="sm")
                    ex4_btn = gr.Button("🚫 Out-of-Scope Test (No Proof)", size="sm")

                with gr.Column(scale=2):
                    chatbot = gr.Chatbot(label="Academic Assistant Chat", height=450)
                    with gr.Row():
                        msg_input = gr.Textbox(
                            placeholder="Ask a question from your course materials...",
                            scale=4,
                            show_label=False,
                        )
                        send_btn = gr.Button("Ask 🚀", variant="primary", scale=1)

                with gr.Column(scale=1):
                    sources_panel = gr.Markdown(
                        value="### 📚 Verified Source Attribution\nAsk a question to view exact file names, page numbers, and chunk IDs."
                    )

            send_btn.click(
                fn=gradio_chat_handler,
                inputs=[msg_input, chatbot, course_dropdown, topk_slider, api_key_box],
                outputs=[chatbot, sources_panel],
            ).then(lambda: "", None, msg_input)

            msg_input.submit(
                fn=gradio_chat_handler,
                inputs=[msg_input, chatbot, course_dropdown, topk_slider, api_key_box],
                outputs=[chatbot, sources_panel],
            ).then(lambda: "", None, msg_input)

            clear_btn.click(
                fn=lambda: ([], "### 📚 Verified Source Attribution\nConversation cleared."),
                inputs=None,
                outputs=[chatbot, sources_panel],
            )

            ex1_btn.click(
                lambda: ("What are the three key roles in Scrum?", "SE303_Software_Engineering"),
                None,
                [msg_input, course_dropdown],
            )
            ex2_btn.click(
                lambda: ("Explain Third Normal Form (3NF) in databases.", "DB202_Database_Systems"),
                None,
                [msg_input, course_dropdown],
            )
            ex3_btn.click(
                lambda: ("What is the attendance and midterm exam policy in CS101?", "CS101_Artificial_Intelligence"),
                None,
                [msg_input, course_dropdown],
            )
            ex4_btn.click(
                lambda: ("What is the recipe for pasta carbonara?", "ALL"),
                None,
                [msg_input, course_dropdown],
            )

        with gr.TabItem("📤 Upload Course Material"):
            gr.Markdown("### Add New Documents to ChromaDB (`PDF`, `DOCX`, `CSV`, `TXT`)")
            with gr.Row():
                file_input = gr.File(label="Select Course File")
                upload_course_dropdown = gr.Dropdown(
                    choices=AVAILABLE_COURSES,
                    value=AVAILABLE_COURSES[0],
                    label="Assign to Course",
                )
            upload_btn = gr.Button("⚙️ Process & Index Document", variant="primary")
            upload_status = gr.Markdown("")

            upload_btn.click(
                fn=handle_file_upload,
                inputs=[file_input, upload_course_dropdown],
                outputs=upload_status,
            )

if __name__ == "__main__":
    demo.launch(
        share=False,
        theme=gr.themes.Soft(primary_hue="blue", secondary_hue="slate"),
    )
````
